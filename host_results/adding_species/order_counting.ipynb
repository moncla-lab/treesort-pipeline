```python
import pandas as pd
import os
import sys
import numpy as np
import re

module_dir = "/Users/monclalab1/Documents/scripts/"
sys.path.append(module_dir)

from fasta_editing import fasta_to_df, fasta_writer
```

```python
fasta_file = "fastas/h3nx_ha_master.fasta"
species = "species.csv"
```

```python
df = fasta_to_df(fasta_file)
    
df['strain'] = df['header'].str.lower().str.split("|").str[0]
df['order1'] = df['header'].str.lower().str.split("|").str[4]
```

```python
df_avian = df[df['order1'] == 'avian'].copy()
df_avian['strain'] = df_avian['header'].str.lower().str.split("|").str[0]
df_avian['species'] = df_avian['strain'].str.lower().str.split("/").str[1]
df_avian = df_avian.dropna(subset=['species'])
```

```python
species = pd.read_csv("classifications/species.csv")
species.annotated = species.annotated.str.lower()
species.order = species.order.str.lower()
```

```python
merged = df_avian.merge(
    species[['annotated', 'broad', 'order']],
    left_on='species',
    right_on='annotated',
    how='left'
)
```
