```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.lines as mlines
import seaborn as sns

import warnings

warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
segment_logs = "logs-segments"
clades = ["h3nx", "canine", "equine", "human", "avian", 'swine']
```

```python
segment_cols = ["NA_rate", "NP_rate", "NS_rate", "PB1_rate", "PB2_rate", "MP_rate", "PA_rate"]
colors = ['#fee391', '#c7e9b4', '#7fcdbb', '#41b6c4', '#1d91c0', '#225ea8', '#0c2c84']

for clade in clades:
    
    df = pd.read_csv(f"{segment_logs}/{clade}.csv")
    
    long_df = df.melt(
        id_vars="rep",
        value_vars=segment_cols,
        var_name="segment",
        value_name="rate"
    )

    long_df["segment"] = long_df["segment"].str.replace("_rate", "")

    # anova
    from scipy.stats import f_oneway
    
    groups = [group["rate"].dropna().values 
              for seg, group in long_df.groupby("segment")]

    anova_result = f_oneway(*groups)

    print(f"{clade} ANOVA:")
    print(f"  F = {anova_result.statistic:.4f},  p = {anova_result.pvalue:.4e}")
    print("-" * 40)
    
    # paired t-test
    from scipy.stats import ttest_rel
    import itertools
    import statsmodels.stats.multitest as smm

    wide = long_df.pivot(index="rep", columns="segment", values="rate")
    pairs = list(itertools.combinations(wide.columns, 2))

    raw_pvals = []
    results = []

    for s1, s2 in pairs:
        t, p = ttest_rel(wide[s1], wide[s2], nan_policy='omit')
        raw_pvals.append(p)
        results.append((s1, s2, t, p))

    # bonferonni correction
    reject, pvals_corr, _, _ = smm.multipletests(raw_pvals, method="bonferroni")

    print(f"{clade} paired t-tests (bonferroni-corrected):")
    for (s1, s2, t, p_raw), p_corr, r in zip(results, pvals_corr, reject):
        print(f"  {s1} vs {s2}: t={t:.3f}, raw p={p_raw:.3e}, corrected p={p_corr:.3e}, significant={r}")
    print("-" * 60)
    
    
    # plot
    
    palette = dict(zip([s.replace("_rate", "") for s in segment_cols], colors))

    plt.figure(figsize=(12, 6))

    sns.violinplot(
        x="segment", y="rate",
        data=long_df, inner=None,
        palette=palette
    )

    sns.stripplot(
        x="segment", y="rate",
        data=long_df,
        color="dimgrey",
        size=3, jitter=True, alpha=0.6
    )

    plt.xticks(rotation=45)
    ymax = long_df["rate"].max()
    plt.ylim(0, ymax + 0.02)
    plt.title(clade)
    plt.tight_layout()
    plt.show()
```
